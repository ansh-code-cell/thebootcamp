```python
import random

numbers = [random.randint(1, 100) for i in range(20)]

print("Random numbers:", numbers)

print("Even numbers:")
for num in numbers:
    if num % 2 == 0:
        print(num)
```

```python
text = "Python is very interesting"

print("Original string:", text)

print("String in reverse order:", text[::-1])

words = text.split()
print("Words in reverse order:", " ".join(words[::-1]))
```
